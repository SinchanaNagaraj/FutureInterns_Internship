```python
from collections import defaultdict
import random

def build_markov_chain(words):
    chain = defaultdict(list)
    for w1, w2 in zip(words[:-1], words[1:]):
        chain[w1].append(w2)
    return chain

def generate_text(chain, length=50, seed=None):

    word = seed or random.choice(list(chain))
    output = [word]

    for _ in range(length - 1):
        next_words = chain.get(word)
        if next_words:
            word = random.choice(next_words)
        else:
            
            word = random.choice(list(chain))
        output.append(word)

    return ' '.join(output)

def main():
    with open("TRAVELLER.txt", "r", encoding="utf-8", errors="ignore") as f:
        text = f.read().lower()
    words = text.split()

    chain = build_markov_chain(words)
    generated = generate_text(chain, length=100)
    print(generated)

if __name__ == "__main__":
    main()
```
